```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense,Input, LSTM, GRU
```

```python
# load the imdb dataset
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)

# Pad sequences to have the same length
X_train = pad_sequences(X_train, maxlen=100)
X_test = pad_sequences(X_test, maxlen=100)
```

```python
# define the RNN model :-

model = Sequential()

# Add this Input layer first
model.add(Input(shape=(100,)))

model.add(Embedding(10000, 32))                        # Embedding layer to convert words to vectors
model.add(SimpleRNN(5, return_sequences=True))         # RNN layer with 5 units
model.add(SimpleRNN(5))                               # Another RNN layer with 5 units
model.add(Dense(1, activation='sigmoid'))             # Output layer for binary classification


model.summary()
```

```python
# compile the model

model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# train the model 

model.fit(X_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
# Similarly we can also define the LSTM model and GRU model 
```

```python
# Define the LSTM model

model_2 = Sequential()

# Add this Input layer first
model_2.add(Input(shape=(100,)))

model_2.add(Embedding(10000, 32))                 # Embedding layer to convert words to vectors
model_2.add(LSTM(5, return_sequences=True))         # Add LSTM layer with 5 units
model_2.add(LSTM(5))                               # Another LSTM layer with 5 units
model_2.add(Dense(1, activation='sigmoid'))             # Output layer for binary classification


model_2.summary()
```

```python
# compile the model_2

model_2.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# train the model 

model_2.fit(X_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```

```python
# Define the GRU model

model_3 = Sequential()

# Add this Input layer first
model_3.add(Input(shape=(100,)))

model_3.add(Embedding(10000, 32))                 # Embedding layer to convert words to vectors
model_3.add(GRU(5, return_sequences=True))         # Add GRU layer with 5 units
model_3.add(GRU(5))                               # Another GRU layer with 5 units
model_3.add(Dense(1, activation='sigmoid'))             # Output layer for binary classification


model_3.summary()
```

```python
# compile the model_3

model_3.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
# train the model 

model_3.fit(X_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```
